```python
layer_nodes = []
for i in range(11):
    distance = i
    nodes = set(nx.ego_graph(net, 'root', radius=distance))
    # remove nodes that are not **at** the specified distance but closer
    if distance >= 1.:
        nodes -= set(nx.ego_graph(net, 'root', radius=distance - 1))
    layer_nodes.append(nodes) 
```

## 5.2 Tissue Maps 

```python
tissue_name = 'Prostate'
prostate_maps = tissue_maps[tissue_name]
```

```python
# get edges from maps 
def get_edges(maps, gene_included=False):
    child = []
    parent = []
    if gene_included: x = 0 
    else: x = 1  
        
    for idx, mapp in enumerate(maps[x:-1]):
        new_child = mapp.index[mapp.sum(axis=1).values.astype(bool)].tolist()
        for jdx, c in enumerate(new_child):
            new_parent = mapp.columns[mapp.loc[c].values.astype(bool)]
            for p in new_parent:
                child.append(c)
                parent.append(p)

    edge_df = pd.DataFrame({'child': child, 'parent': parent})
    return edge_df
```

```python
maps_edge = get_edges(prostate_maps)
```

```python
# Create a directed graph
G = nx.DiGraph()

# Add edges to the graph
edges = [(row['parent'],row['child']) for _, row in maps_edge.iterrows()]
G.add_edges_from(edges)
```

```python
roots = [n for n, d in G.in_degree() if d == 0]
root_node = 'root'
edges = [(root_node, n) for n in roots]
G.add_edges_from(edges)
```

```python
layer_nodes = []
for i in range(11):
    distance = i
    nodes = set(nx.ego_graph(G, 'root', radius=distance))
    # remove nodes that are not **at** the specified distance but closer
    if distance >= 1.:
        nodes -= set(nx.ego_graph(G, 'root', radius=distance - 1))
    layer_nodes.append(nodes) 
```

```python
layer_num = [len(x) for x in layer_nodes]
layer_num
```

```python
reactome_id=[]
distance_from_root=[]
node_gene_num=[]
for ldx, l in enumerate(layer_nodes):
    for n in l:
        reactome_id.append(n)
        distance_from_root.append(ldx)
        node_gene_num.append(len(gene_pathway.loc[gene_pathway.reactome_id == n,'gene']))
```

```python
df = pd.DataFrame({'reactome_id':reactome_id,
                   'distance_from_root':distance_from_root,
                   'node_gene_num':node_gene_num})
df = df.drop(0)
```

```python
maps_node = pd.merge(df, pathways, on='reactome_id', how='left')  # 'inner' merge keeps common rows
```

```python
maps_edge.to_csv('../../Cytoscape_data/Prostate_maps_edge.csv')
maps_node.to_csv('../../Cytoscape_data/Prostate_maps_node.csv')
```
