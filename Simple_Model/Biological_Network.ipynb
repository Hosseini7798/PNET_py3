# 1. Reactome 

```python
import pandas as pd 
import numpy as np 
from network.gmt_reader import GMT
```

```python
# https://reactome.org/download/current/ReactomePathways.txt
# Pathways: reactome_id, pathway_name, species
pathways = pd.read_csv('../_database/pathways/Reactome2023/ReactomePathways.txt',sep='\t')
pathways.columns = ['reactome_id', 'pathway_name', 'species']
pathways = pathways[pathways.species == 'Homo sapiens'].reset_index(drop=True)
pathways.drop('species', inplace=True,axis=1)

# https://reactome.org/download/current/ReactomePathwaysRelation.txt
# hierarchy: parent, child
hierarchy = pd.read_csv('../_database/pathways/Reactome2023/ReactomePathwaysRelation.txt', sep='\t')
hierarchy.columns = ['parent', 'child']
hierarchy = hierarchy[hierarchy['child'].str.contains('HSA')].reset_index(drop=True)

# https://reactome.org/download/current/ReactomePathways.gmt.zip
# gene_pathway: reactome_id, gene
gmt = GMT()
gene_pathway = gmt.load_data('../_database/pathways/Reactome2023/ReactomePathways.gmt', pathway_col=1, genes_col=3)
gene_pathway.columns = ['reactome_id', 'gene']

# https://reactome.org/download-data#:~:text=Complexes%20to%20Pathways%20(Human)
# complex_pathway: complex, pathway, top_level_pathway
complex_pathway = pd.read_csv('../_database/pathways/Reactome2023/Complex_2_Pathway_human.txt',sep='\t')
complex_pathway = complex_pathway[complex_pathway['complex'].str.contains('HSA')].reset_index(drop=True)

# https://reactome.org/download/current/Ewas2Pathway_human.txt
# Ewas: ewas, pathway, top_level_pathway
Ewas = pd.read_csv('../_database/pathways/Reactome2023/Ewas2Pathway_human.txt',sep='\t')
Ewas = Ewas[Ewas['ewas'].str.contains('HSA')].reset_index(drop=True)

# https://reactome.org/download/tools/ReatomeFIs/FIsInGene_070323_with_annotations.txt.zip
# FIS: Gene1, Gene2, Annotation, Direction, Score
FIS = pd.read_csv('../_database/pathways/Reactome2023/FIsInGene_070323_with_annotations.txt',sep='\t')
```
